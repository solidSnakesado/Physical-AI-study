- 비즈니스 데이터를 활용한 EDA 학습
    2. 비즈니스 데이터 이해하기
        - 학습 목표
            - 비즈니스 환경에서 사용되는 다양한 데이터 유형을 이해한다.
            - 데이터 품질 평가 방법과 개선 프로세스를 학습한다.
            - 비즈니스 컨텍스트와 데이터 분석을 연결하는 방법을 익힌다.
            - 효과적인 분석 질문과 가설을 설정하는 프레임워크를 습득한다.
        - 비즈니스 데이터 유형
            - 정형 데이터
                - 정의: 정해진 스키마를 따르는 데이터
                - 예시: 관계형 DB, 스프레드 시트
                - 처리 난이도: 쉬움
                - 분석 도구: SQL, Excel, Tableau, Python
            - 반정형 데이터
                - 정의: 일부 구조화된 데이터
                - 예시: JSON, XML, 로그파일
                - 처리 난이도: 보통
                - 분석 도구: Python, NoSQL
            - 비정형 데이터
                - 정의: 정해진 형식이 없는 데이터
                - 예시: 텍스트, 이미지, 영상, 음성
                - 처리 난이도: 어려움
                - 분석 도구: Python, NLP, 컴퓨터 비전
            - 정형 데이터의 장점
                - 표준화된 분석 방법 적용 가능
                - 빠른 쿼리와 집계 연산
                - 다양한 분석 도구 지원
            - 비정형 데이터의 활용
                - 풍부한 인사이트 제공
                - 고객 감정 분석 가능
                - 실시간 트렌트 파악
                - 새로운 비즈니스 기회 발견
        - 데이터 품질 평가
            - 데이터 품질의 6가지 차원
                - 정확성 (Accurary)
                    - 정의: 데이터가 실제 현시을 올바르게 반영하는 정도
                    - 예시: 고객의 주소가 실제 주소와 일치
                    - 측정: 다른 유사한 데이터와 비교
                - 완전성 (Completeness)
                    - 정의: 필요한 모든 데이터가 누락없이 존재하는 정도
                    - 예시: 이메일 주소 누락률 5% 미만
                    - 측정: 결측치 비율 확인
                - 일관성 (Consistency)
                    - 정의: 동일한 데이터가 여러 시스템에 같은 값을 가지는 정도
                    - 예시: CRM과 ERP 시스템 고객 정보 일치
                    - 측정: 시스템간 데이터 비교
                - 유효성 (Validity)
                    - 정의: 데이터가 정의된 형식과 규칙을 따르는 정도
                    - 예시: 이메일, 전화번호 형식 준수
                    - 측정: 형식 검증, 비즈니스 규칙 검사
                - 교유성 (Uniqueness)
                    - 정의: 중복 없이 각 데이터가 유일하게 존재하는 정도
                    - 예시: 고객 ID가 중복되지 않음
                    - 측정: 중복 레코드 탐지, 유일성 제약 검사
                - 적시성 (Timeliness)
                    - 정의: 비즈니스 요구사항에 맞는 시점에 이용 가능한 정도
                    - 예시: 실시간 재고 5분 이내 업데이트
                    - 측정이 데이터 생성-이용 간격 측정
            - 데이터 품질 이슈 해결 프로세스
                1. 기초 통계 분석
                    - 필드별 기본 통계량 (평균, 중앙값, 표준편차)
                    - 타입과 형식 확인
                    - 값의 분포와 빈도 분석
                2. 품질 이슈 확인
                    - 결측치 패턴 분석
                    - 이상치 탐지 (통계적 방법, 도메인 규칙)
                    - 중복 데이터 식별
                    - 형식 위반 사례 확인
                3. 영향도 평가
                    - 비즈니스 영향 평가
                    - 분석 영향도 확인
                    - 품질 이슈 발생별 우선순위 설정
                4. 해결 방안 수립
                    - 임시 해결 방안
                    - 근본적 해결 방안
                    - 예방 체계 구축
        - 비즈니스 컨텍스트에서 데이터 이해하기
            - 비즈니스 컨텍스트 = 왜 도메인 지식이 중요한가
            - 비즈니스 별로 데이터 의미, 중요성, 실행 가능성이 다름
                - 데이터 VS. 도메인 지식
                    - 게임 산업 데이터 분석
                        - 데이터 수집: 회원가입율 84%, 위치정도 의미 X
                        - 상품 구매 시간대: 평일 20시 ~ 01시 구매 70%
                        - 사용 시간 (play 시간): 일 평균 1.2 시간
                        - 학십 지표: 재방문율, play 시간
                    - 커피업계 데이터 분석
                        - 회원 데이터 수집: 회원 가입율 30%, 위치 중요
                        - 상품 구매 시간대: 평일 07시 ~ 20시 구매 70%
                        - 매장 사용 시간: 측정 불가
                        - 핵심 지표: 재구매율, 구매 상품수
                    - 도메인 전문가와 협업 방법
                        - 효과적인 질문 기법
                            - 이 지표가 증가/감소했을 때 실제로 어떤 일이 일어나나요?
                            - 과거에 비슷한 패턴을 본적이 있나요?
                            - 이 데이터에 영향을 주는 외부 요인은 무엇인가요?
                        - 데이터 해석 검증
                            - 이 분석 결과가 현실적으로 가능한가요?
                            - 예상과 다른 결과인데, 어떤 이유가 있을까요?
                            - 이 인사이트를 실행하기 어려운 제약 사항이 있나요?
            - 비즈니스 KPI와 데이터 변수 간의 관계 이해
                - 재무 성과 KPI
                    - 매출 (Revenue): 주문 금액, 결제 완료 건수
                    - 수익성 (Profitability): 원거, 마케팅 비용, 운영비
                    - 현금흐름 (Casj Flow): 입금/출금 데이터, 미수금
                - 고객 관련 KPI
                    - 고객 획득 비용 (CAC): 마케팅 비용 / 신규 고객수
                    - 고객 생애 가치 (LTV): 평균 구매금이개 x 구매 빈도 x 고객 수명
                    - 고객 만족도 (CSAT): 설문 점수, 리뷰 평점, 재구매율
                - 운영 효율성 KPI
                    - 전환율 (Conversion Rate): 방문자 수 대비 구매 고객수
                    - 재고 회전율 (Inventory Turnover): 매출원가 / 평균 재고
                    - 배송 만족도: 배송 시간, 정확도, 고객 피드백
                - 고객 이탈율
                    - 고객 이탈율 = (기간내 미사용 고객수 / 기간내 총 고객수) x 100
                    - 최근 1년 이탈률 4% (업게 평균 5.2%)
                    - 월별 이탈률 1% 증가 -> 월매출 12억 감소
                    - VIP 이탈률 2.1%로 전년대비 0.4%p 감소
            - 질문 및 가설 만들기
                - 올바른 질문이 성공적인 분석의 출발점
                - 질문 유형별 세분화
                    - 탐색적 질문 (Exploratory)
                        - "무엇이?"
                        - 현상을 이해하고 원인을 파악
                        - "무엇이 재구매율을 높이는가?", "우리 고객의 구매 퍄턴은 어떻게 되는가?"
                    - 예측적 질문 (Predictive)
                        - "언제? 얼마나?"
                        - 미래 상황을 예측
                        - "1+1 이벤트는 얼마나 구매를 증가시킬수 있을까?", "다음달 신재품 판매량은 얼마나 될까?"
                    - 처방적 질문 (Prescriptive)
                        - "어떻게?"
                        - 최적의 행동 방안 찾기
                        - "이탈률을 10% 줄이려면 어떻게 해야할까?", "어떻게 하면 매출을 높일 수 있을 까?"
                - 질문 개선 예시
                    - 나쁜 질문
                        - "마케팅이 효과적인가?"
                            - 대상 모호함: 어떤 종류의 마케팅?, 무엇이 효과적인지?
                            - 측정 불가: 효과를 평가하는 기준이 없음
                            - 시한 없음: 언제부터 마케팅인지 확인 불가
                    - 좋은 질문
                        - "지난 6개월간 고객 정보를 입력한 신규고객은 그렇지 않은 고객 보다 30일내 이탈률이 10%이상 낮은가?"
                            - 구제척: 고객 정보 입력 고객 vs 입력하지 않은 고객
                            - 측정 가능: 30일내 이탈률, 10% 기준
                            - 시한의 명확: 최근 6개월간 가입고객
                    - 핵심 포인트
                        - 질문의 품질이 분석의 품질을 결정합니다 - 올바른 질문 설정이 중요
                        - 구체적일 수록 실행 가능한 인사이트를 얻습니다 - 모호한 질문은 모호한 답을 낳음
                        - 비즈니스 목표와 연결괸 질문이 가치를 창출합니다 - 의사결정에 도움이 되는 질문
                        - 측정 가능한 지표로 표현해야 검증이 가능합니다 - 주관적 판단보다 객관적 지표
                - 비즈니스 문제를 검증 가능한 데이터 문제로 바꾸는 핵심
                - 가설은 우리가 가지고 있는 데이터로 수치화하여 검증한 형식이어야 합니다.
                    - 변환 프로세스
                        - 비즈니스 문제: "지난 6개월간 고객 정보를 입력한 신규고객은 그렇지 않은 고객보다 30일내 이탈율이 10%이상 낮은가?"
                        - 데이터 문제: "가입시 고객 정보를 입력한 고객집단과 그렇지 않은 고객 집단은 평균 이탈율이 차이가 없을 것이다."
                    - 통계적 가설 검증의 기본 형태
                        - 귀무 가설(H0) <- 이 부분이 중요, 이 가설이 성립하지 않으면 대립으로 진행
                            - "차이가 없다" / "관계가 없다"
                                - 두 집단간 평균의 차이가 없다
                                - 두 변수간 상관관계가 없다
                                - 이벤트 효과가 없다
                        - 대립 가설(H1)
                            - "차이가 있다" / "관계가 있다"
                                - 두 집단간 평균의 차이가 있다
                                - 두 변수간 상관관계가 있다.
                                - 이벤트 효과가 있다.
                - 어느 가설을 선택할지 기준? - 유의미한 차이가 있는 지 검증은 통계적 검정 사용
                    - 데이터로 검증하기 쉬운 가설 형태
                        - 집단간 비교
                            - "A그룹과 B그룹의 평균은 차이가 없다" -> t-검정, ANOVA
                        - 변수간 관계
                            - "변수X와 변수Y 사이에 관계가 없다" -> 상관분석, 회귀분석
                        - 비율 / 분포 비교
                            - "두 집단의 비율에 차이가 없다" -> 카이제곱 검정
                    - 실무 적용 팁
                        1. 측정 가능한 숫자 지표로 변환 - "고객이 만족한다" -> 만족도는 NPS점수로 평가
                        2. 비교 가능한 집단 설정 - "기존고객" vs "신규고객"처엄 명확한 집단을 구분
                        3. 귀무가설을 먼저 설정 - "차이가 없다"부터 시작하여 반즈을 찾는 방식
                        4. 효과 크기도 고려 - 통계적 유의성과 실무적 차이의 크기를 함께 판단
    3. 주요 EDA 1 - 단변량 변수 분석
        - 학습 목표
            - 연속형 변수의 특성을 파악하는 다양한 방법 학습
            - 범주형 변수의 분포와 불균형 문제 해결법 이해
            - 시계열 데이터의 패턴 분석 기법 습득
            - 적절한 시각화 방법 선택과 해석 능력 향상
        - 연속형 변수 분석
            - 데이터의 중심을 찾아라!
                - 평균(Mean)
                    - 계산: 모든 값의 합계 / 계수
                    - 장점: 직관적, 계산 쉬움
                    - 단점: 이상치에 민감
                    - VIP 고객의 영향을 받는 값
                - 중앙값 (median)
                    - 계산: 전체를 정렬했을때 가운데 값
                    - 장점: 이상치에 강함
                    - 단점: 극단 정보 손실
                    - 일반 고객들의 실제 구매 수준
                - 최빈값 (Mode)
                    - 계산: 가장 자주 나타난 값
                    - 특징: 범주형 데이터 유용
                    - 가장 많이 구매하는 금액대
            - 실무 팁: 평균과 중앙값이 크게 다르면 이상치를 의심해 봐야 함
            - 데이터가 얼마나 흩어져 있을까?
                - 표준편차 (Standard Deviation): 평균으로 부터 평균적으로 얼마나 떨어져 있는가?
                    - 원래 단위와 동일
                    - 작을 수록 -> 모여 있음
                - 분산 (variance): 표준 편차의 제곱
                    - 수학적 계산이 표준편차보다 유리
                - 범위 (Range): 최대값 - 최소값
                    - 간단하지만 계산
                    - 이상치에 매우 민감
                - 사분위범위 (IQR): Q3(75%) - Q1(25%) (3분위 수에서 1분위 수를 뺀것)
                    - 이상치에 강함, 박스플롯의 핵심
