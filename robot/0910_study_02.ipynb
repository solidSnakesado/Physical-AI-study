- AICE : 인공지능 활용능력을 평가하는 AI 자격증 시험
    - KT가 개발
    - 3가지 AICE 요소: 빅데이터 해석 / 인공지능 기술 - 상황 / 현실성 있는 문제 해결
- AICE 종류
    - PROFESSIONAL: 현직 개발자 / 전공자
    - ASSOCIATE: 준전공자   <- 이것만 공인 자격증
    - BASIC, JUNIOR, FUTURE(1, 2, 3급): 비전공자
- 특징
    - 현실성 있는 문제를 인공지능(AI)으로 얼마나 잘 해결 -> 실제 사례 기반 - 경험
    - 총 14문항(코딩 작성) / 파이썬 언어 사용
    - 응시료 80,000 / 시험 시간 90분(오프라인 시험 존재 / 보통은 온라인 시험)
        - 노트북(캠) / 데트크탑의 경우도 캠 준비 + 30분 전 캠 ON - 검토 / 감독관 승인
    - 합격 점수는 100점 만점 - 80점 이상
    - 출제 기분: 크게 4가지로 분류
        - 탐색적 데이터 분석 / 데이터 전처리 / 머신러닝, 딥러닝 / 모델 성능 평가
