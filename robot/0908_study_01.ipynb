아누이노 프로그래밍 기초
 - pinMode() 함수의 경우 디지털 신호에만 필요, 출력일지 입력일지 지정하는 것인데, 아날로그는 기본적으로 둘다 지원
 - pinMOde: 지정된 핀의 동작을 입력 혹은 출력으로 지정하기 위한 함수, Digital IO는 필수
 - millis() 함수 설명 체크 필요
 - Serial
    - 유선으로 연결된 상태에서 주로 사용
    - Serial.begin(9600)    // Serial Port Open & 통신속도 설정(초당 비트수)

- 아두이노 데이터 타입
    - 기본 타입 (논리형, 문자형, 바이트)
        | 데이터 타입 | 8비트 크기 | 8비트 범위 | 32비트 크기 | 32비트 범위 | 주요 용도 | 
        | --- | --- | --- | ---- | --- | --- |
        | boolean / bool | 1바이트 | true/false | 1바이트 | true/false | 논리값, 조건 플래그, 상태 관리 |
        | char | 1바이트 | -128~127 | 1바이트 | -128~127 | 문자 저장, 직은 정수 값, ASCII 코드 |
        | byte(unsigned char) | 1바이트 | 0~255 | 1바이트 | 0~255 | 바이트 데이터, RGB 값, 비트 연산 |

    - 부호 있는 정수형
        | 데이터 타입 | 8비트 크기 | 8비트 범위 | 32비트 크기 | 32비트 범위 | 주요 용도 | 
        | --- | --- | --- | ---- | --- | --- |
        | short | 2바이트 | -32,768~32,767 | 2바이트 | -32,768~32,767 | 메모리 절약용, 센서 오프셋 값 |
        | int | 2바이트 | -32,768~32,767 | 4바이트 | -2,147,483,648~2,147,483,647 | 일반 연산, 카운터, 센서 값 저장 |
        | long | 4바이트 | -2,147,483,648~2,147,483,647 | 4바이트 | -2,147,483,648~2,147,483,647 | 시간 차 계산, 누적 합계, 큰 정수 연산 |

    - 부호 없는 정수형
        | 데이터 타입 | 8비트 크기 | 8비트 범위 | 32비트 크기 | 32비트 범위 | 주요 용도 | 
        | --- | --- | --- | ---- | --- | --- |
        | unsigned int | 2바이트 | 0~65,535 | 4바이트 | 0~4,294,967,295 | 카운터, 타이머 값, 양수 연산 |

    - 실수형 (부동소수점)
        | 데이터 타입 | 8비트 크기 | 8비트 범위 | 32비트 크기 | 32비트 범위 | 주요 용도 | 
        | --- | --- | --- | ---- | --- | --- |

    - 데이터 타입이 중요한 이유
        - 오버플로 방지: 올바른 크기와 범위의 데이터 타입 선정
        - 오차 방지: 서로 다른 데이터 타입의 연산, 혹은 잘못된 데이터 타입 변환은 오차를 발생, 명시적인 사용이 필요
