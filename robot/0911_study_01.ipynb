- 학습 목표
    - EDA의 정의와 목적을 명확히 이해한다.
    - 비즈니스 의사결정에서 EDA의 역할을 파악한다.
    - 체계적인 EDA 프로세스를 습득한다.
    - 데이터에서 의미 있는 인사이트를 도출하는 방법을 학습한다.

- EDA 의 정의와 목적
    - Exploratoruy Data Analysis
    - 데이터를 있는 그대로 탐색하여 숨겨진 패턴과 특성을 반견하는 과정
    - 역사적 배경
        - John Tukey (1977)가 최초 제안
        - 탐정처럼 데이터를 조사하라
        - 기존 확인적 분석에서 탐색적 분석으로 패러다음 전환
    - EDA의 목적
        - 데이터의 전체적인 구조 파악
        - 이상치와 결착치 탐지
        - 변수 간 관계 발견
        - 가설 생성 및 검증 방향 설정
    - 기술통계 와 EDA의 차이점
        - 기술통계(Descriptive Statistics)
            - 목적: 데이터를 요약하고 기술
            - 접근방식: 수치적 요약 (평균, 표준편차)
            - 결과물: 통계량 테이블
            - 사고방식: "데이터가 어떻게 생겼나?"
        - 탐색적 데이터 분석 (EDA)
            - 목적: 패턴과 관계를 탐색하고 발견
            - 접근방식: 시각화 + 수치 + 직관
            - 결과물: 그래프, 차트, 인사이트
            - 사고방식: "데이터가 무엇을 말하나?"
