- 딥러닝은 인간의 뇌 신경마늘 학습하는 방식에서 영감을 얻어 고안
- 두뇌의 신경세포가 연결된 형태를 모방하여 인공 신경망 모델을 만들고 인공신경망을 학습(Learning)모델로 사용하는 머신러닝 패러다임이 딥러닝(Deep Learning)

- 인공신경망
  - 두뇌의 신경망 정보처리 구조를 모방하여 만든 컴퓨터 계산 알고리즘이 인공신경망
  - 생물학적 뉴런을 인공 뉴런으로 모델링하면 입력에 가중치를 각각 곱하고 이를 모두 더해 활성화 함수를 통과 시키면 출력이 됨
  - 인공신경망 입력층, 은닉층, 출력층으로 구성

- 기중치와 편향 이해하기
  - 인공신경망 모델을 수학적으로 표현하면 아래 그림과 같음
  - 
  - 뉴런에 2개의 입력값 x1, x2 입력, 입력 개수만큼 2개의 가중치(weight) w1, w2와 1개의 편향(bias, 바이어스) b를 가짐
  - 가중치는 입력 데이터에서 원하는 연산 결과를 만들도록 비중을 부여하는 값, 편향은 뉴런의 활성화를 조절하는 상수
  - 가중치와 편향은 뉴런의 동작 특성을 나타내는 중요한 파라미터로서 이 값들을 조정해 원하는 출력값을 만들어냄
  - 인공신경망으로 실제 정답에 근사하는 출력값을 만들기 위해 뉴런의 가중치와 편향을 반복적으로 조정하며, 이러한 반복적인 과정을 학습이라고 함
  - 
  - 입력 x1이 매우 큰 값이라면 x1이 연산이 미치는 정도는 절대적일 것 하지만 x1에 곱하는 가중치(w1)가 아주 작다면 x1 * w1 의 연산 결과는 0에 가까운 값이됨, 즉 가중치는 입력 데이터의 연살 결과에 미치는 영향도를 조절하는 요소이고, 편향(b)도 가중합을 조정하는 파라미터  

- 활성화 함수 알아보기
  - 인공신경망 뉴런에서는 입력된 신호가 특정 강도 이상일때 다음 뉴런으로 신호를 전달하는 역할을 수행하는 활성화 함수가 있음
  - 활성화 함수는 입력값들의 수학적 선형 결과인 가중합 값을 입력받고 일정 기준에 따라 변환하여 값을 출력하는 비선형(또는 선형) 함수
  - 주요 활성화 함수는 시그모이드(sigmoid), 하이퍼보릭 탄젠트(hyperbolic tangent), 렐루(ReLU), 계산이 가벼운 렐루가 많이 사용

  - 시그모이드 미분 함수 f'(x)를 그래프로 나타내면 다음과 같음
  - 시그모이드 함수는 입력값을 비선형 형태 0~1 사이의 값으로 변화시키므로, 로지스틱 회귀와 같은 이진 분류의 결과를 확률적으로 나타내는 데 사용
  - 시그모이드 미분함수 f'(x)의 값은 입력값이 0일때 0.25에 불과하고 입력 x가 매우 작아지거나 커지면 기울기(미분값)가 0에 수렴
  - 인공신경망을 사용하는 딥러닝은 모델 학습에 활성화 함수의 기울기(gradient, 그래디언트)를 계속 곱하는 연산이 있으며 미분값이 작은 시그모이드 함수를 인닉픙의 활성화 함수로 사용하면 기울기 소실(vanishing gradient) 문제가 발생해 모델 학습이 제대로 이루어지지 않는 문제가 발생

- 하이퍼볼릭 탄젠트(hyperbolic tangent, tanh)
  - 시그모이드 활성화 함수를 사용하는 경우 은닉층이 많아질수록 효과적으로 학습되지 않는 기울기 소실 문제를 개선하기 위한 방법의 하나로 제안
  - 하이퍼볼릭 탄젠트 함수의 출력은 -1과 1사이의 값이고, 미분값이 시그모이드 함수에 비해 커짐
  - 하지만 하이퍼볼릭 탄젠트 함수고 x 값이 크거나 작아짐에 따라 기울기가 매우 작아져서 기울기 소실 문제가 발생

  - 렐루 함수는 입력값이 양수일 경우 출력값은 입력값과 같고, 음수일경우 출력값이 0 으로 식은 아래와 같음
