### 5. 이상치 처리  <sub>(코드 작성)</sub>

df에서 residual_sugar 값이 40을(를) 초과하는 행을 이상치로 보고 삭제한 결과를 df_temp에 저장하세요.

<sub>앞 문항을 못 풀었다면 먼저 `setup(5)` 실행</sub>

```python
df_temp = df[df["residual_sugar"] <= 40]
```

```python
check(5)
```

---
### 6. 결측치 처리  <sub>(코드 작성)</sub>

df_temp의 residual_sugar 컬럼 결측치를 해당 컬럼의 중앙값으로 채운 뒤, 남은 결측치가 있는 행을 삭제하여 df_na에 저장하세요.

<sub>앞 문항을 못 풀었다면 먼저 `setup(6)` 실행</sub>

```python
df_temp["residual_sugar"] = df_temp["residual_sugar"].fillna(df_temp["residual_sugar"].median())
df_na = df_temp.dropna()
```

```python
check(6)
```

---
### 7. 컬럼 삭제·선택  <sub>(코드 작성)</sub>

axis 파라미터를 사용해 df_na에서 wine_id 컬럼을 삭제하고 결과를 df_del에 저장하세요.

<sub>앞 문항을 못 풀었다면 먼저 `setup(7)` 실행</sub>

```python
df_del = df_na.drop(columns="wine_id", axis=1)
df_del[]
```

```python
check(7)
```

---
### 8. 범주형 인코딩  <sub>(코드 작성)</sub>

df_del의 범주형 컬럼 ['type']를 원핫 인코딩하고, 결과를 0/1 정수로 만들어 df_preset에 저장하세요.

<sub>앞 문항을 못 풀었다면 먼저 `setup(8)` 실행</sub>

```python
df_preset = pd.get_dummies(df_del, columns=["type"], dtype=int)
df_preset.head()
```

```python
check(8)
```

---
### 9. 데이터 분할  <sub>(코드 작성)</sub>

sklearn.model_selection에서 train_test_split을 불러오고, df_preset에서 타깃 grade를 y로, 나머지 컬럼을 X로 지정하고, 학습:검증 = 8:2로 나누되 grade 클래스 비율이 원본과 같도록 하세요. 결과는 X_train, X_valid, y_train, y_valid에 저장합니다. (random_state=42)

<sub>앞 문항을 못 풀었다면 먼저 `setup(9)` 실행</sub>

```python
from sklearn.model_selection import train_test_split

y = df_preset["grade"]
X = df_preset.drop(columns="grade")

X_train, X_valid, y_train, y_valid = train_test_split(X, y, stratify=y, random_state=42, train_size=0.8, test_size=0.2)
```

```python
check(9)
```

### 10. 스케일링  <sub>(코드 작성)</sub>

sklearn.preprocessing에서 RobustScaler를 불러와 X_train을 학습(fit)시킨 뒤 X_train과 X_valid를 변환하여 각각 같은 변수에 다시 저장하세요. (이상치 영향을 적게 받도록 중앙값·IQR 기준)

<sub>앞 문항을 못 풀었다면 먼저 `setup(10)` 실행</sub>

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

X_train = scaler.fit_transform(X_train)
X_valid = scaler.transform(X_valid)
```

```python
check(10)
```

### 11. 머신러닝 모델  <sub>(코드 작성)</sub>

scikit-learn에서 RandomForestClassifier를 불러와 n_estimators=100, random_state=42로 모델 rf를 만들고 학습 데이터로 학습시키세요. (다중분류 문제)

<sub>앞 문항을 못 풀었다면 먼저 `setup(11)` 실행</sub>

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42)

rf.fit(X_train, y_train)
```

```python
check(11)
```
