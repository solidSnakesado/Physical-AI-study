### 5. 이상치 처리  <sub>(코드 작성)</sub>

df에서 sepal_width이(가) 4.5 미만이면서 species이(가) 결측이 아닌 행만 남겨 df_temp에 저장하세요.

```python
df_temp = df[(df["sepal_width"] < 4.5) & (df["species"].notna())]
```

---
### 6. 결측치 처리  <sub>(코드 작성)</sub>

df_temp의 sepal_width 컬럼 결측치는 해당 컬럼의 평균으로 채우고, petal_length 컬럼에 결측치가 있는 행은 삭제하여 df_na에 저장하세요.

<sub>앞 문항을 못 풀었다면 먼저 `setup(6)` 실행</sub>

```python
df_temp["sepal_width"] = df_temp["sepal_width"].fillna(df_temp["sepal_width"].mean())
df_na = df_temp.dropna(subset="petal_length")
```

```python
check(6)
```

### 7. 컬럼 삭제·선택  <sub>(코드 작성)</sub>

분석에 불필요한 sample_id 컬럼을 df_na에서 삭제하여 df_del에 저장하세요.

<sub>앞 문항을 못 풀었다면 먼저 `setup(7)` 실행</sub>

```python
df_del = df_na.drop(columns="sample_id")
```

```python
check(7)
```

---
### 8. 범주형 인코딩  <sub>(코드 작성)</sub>

df_del의 범주형 컬럼 ['region']를 원핫 인코딩하되, 각 컬럼의 첫 번째 범주 더미는 제외하고 결과를 0/1 정수로 만들어 df_preset에 저장하세요.

<sub>앞 문항을 못 풀었다면 먼저 `setup(8)` 실행</sub>

```python
df_preset = pd.get_dummies(df_del, columns=["region"], drop_first=True, dtype=int)
```

```python
check(8)
```

### 9. 데이터 분할  <sub>(코드 작성)</sub>

sklearn.model_selection에서 train_test_split을 불러오고, df_preset에서 타깃 species를 y로, 나머지 컬럼을 X로 지정하고, 학습:검증 = 7:3으로 나누되 클래스 비율을 유지하세요. 결과는 X_train, X_valid, y_train, y_valid에 저장합니다. (random_state=2024)

<sub>앞 문항을 못 풀었다면 먼저 `setup(9)` 실행</sub>

```python
from sklearn.model_selection import train_test_split

y = df_preset["species"]
X = df_preset.drop("species", axis=1)

X_train, X_valid, y_train, y_valid = train_test_split(X, y, train_size=0.7, test_size=0.3, random_state=2024, stratify=y)
```

### 10. 스케일링  <sub>(코드 작성)</sub>

sklearn.preprocessing에서 StandardScaler를 불러와 X_train을 학습(fit)시킨 뒤 X_train과 X_valid를 변환하여 각각 같은 변수에 다시 저장하세요. (평균 0, 표준편차 1)

<sub>앞 문항을 못 풀었다면 먼저 `setup(10)` 실행</sub>

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
ss.fit(X_train)

X_train = ss.fit_transform(X_train)
X_valid = ss.transform(X_valid)
```

### 13. 딥러닝 모델  <sub>(코드 작성)</sub>

아래 조건으로 딥러닝 모델 model을 만들고 학습시키세요.  
· 입력: X_train의 특성 개수  
· Dense 64 (relu) → Dropout 0.2 → Dense 32 (relu) → 출력층 (다중분류에 맞게, 타깃은 0~2 정수 라벨)  
· compile: optimizer='adam', 문제 유형에 맞는 loss, metrics=['accuracy']  
· 학습: epochs=30, batch_size=16, 검증 데이터 (X_valid, y_valid), 결과는 history에 저장

<sub>앞 문항을 못 풀었다면 먼저 `setup(13)` 실행</sub>

```python
from tensorflow import keras

class MyModel(keras.Model):
    def __init__(self):
        super().__init__()
        self.dense1 = keras.layers.Dense(64, activation="relu")
        self.dropout = keras.layers.Dropout(0.2)
        self.dense2 = keras.layers.Dense(32, activation="relu")
        self.dense3 = keras.layers.Dense(3, activation="softmax")

    def call(self, inputs, training=False):
        x = self.dense1(inputs)
        x = self.dropout(x, training=training)
        x = self.dense2(x)
        x = self.dense3(x)
        return x

model = MyModel()
```
