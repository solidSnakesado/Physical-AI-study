- AI람 무엇인가?
    - AI(Artificial Intelligence)는 머신러닝(Machine Learning)과 딥러닝(Deep Learning)을 포함

- 머신러닝 이해하기
    - 머신러닝(Machine Learning)은 컴퓨터가 다양한 머신러닝 학습 모델을 기반으로 제공된 데이터를 이용해 학습을 수행하고 모델의 성능을 반복적으로 평가하며, 스스로 성능을 향상하는 것을 말함, 제공된 각 데이터별 상관관계와 특성을 찾아내고 결론을 예측할 수 있음
    - 머신러닝은 데이터를 설명할 수 있는 하나의 함수를 찾는 것과 유사
    - 하나의 함수를 만들면 그것을 활용해 사용자가 임의의 x를 입력했을 때 어떤 y가 나올지를 예측할 수 있음
    - 머신러닝도 여러 데이터를 기반으로 하나의 모델을 생성하고 여기에 데이터를 넣으면 예측 결과를 알 수 있는 원리

- 딥러닝 이해하기
    - 딥러닝(Deep Learning)은 머신러닝과는 구분되는 개념이지만, 큰 틀에서는 머신러닝에 포함된 기술, 딥러닝은 인간의 뇌 신경망이 학습하는 방식에서 영감을 얻어, 머신러닝에 신경망 네트워크를 더해 만들어짐
    - 사람의 뇌에 있는 신경망은 시냅스 간 전기신호를 주고 받으며 정보를 전달, 딥러닝은 이 개념을 도입해 레이어(layer)를 겹겹이 쌓고 레이어 내 노드를 구성하여, 각 노드가 그물처럼 연결되어 학습을 진행, 각 노드는 머신러닝 모델 하나의 역할을 수행하며, 데이터를 입력받고 학습하여 가중지 w를 구합, 최종적으로 모든 레이어를 거치면 전체 학습의 성능을 평가하게 되고 오차 역전파를 통해 딥러닝 모델의 성능을 향상하기 위한 가중치 w를 재계산하는 과정을 반복
    - 이러한 학습 방법에서 알 수 있듯이 딥러닝은 머신러닝에 비해 연산이 기하급수적으로 늘어남에 따라 성능은 뛰어나지만 연산에 파ㅣㄹ요한 자원이나 시간이 오래 소요된다는 점을 감안해야 함
