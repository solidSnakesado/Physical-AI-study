- AI 학습 방법 이해하기
    - AI 학습은 크게 지도학습, 비지도학습, 강화학삽 3가지로 분류, 이번엔 지도학습, 비지도학습 설명 진행

- 지도학습 이해하기
    - 지도학습(Supervised Learning)은 정답을 알려주며 학습시키는 방식, 지도학습은 어린아이를 가르치는 것과 같음, 예를 들어 아이에게 동그라미와 세모를 구분하는 방법을 가르치는 과정과 같은 두 도형이 모양이 다르다는 것만 인지하고 있을 때 동그라미와 세모의 정답을 반복해서 얼려주면 학습을 통해 분류 가능
    - 이처럼 지도학습은 문제의 정답이 있어야만 학습을 진행할 수 있음, 정답은 레이블(Label)이라고도 불리며, 데이터에 레이블을 입력해 주는 것을 레이블링(Labeling)이라고 함
    - 지도학습은 크게 회귀(Regression)와 분류(Classification)로 나뉨, AI를 수행하여 확인하고자 하는 결과의 형태가 무엇이냐에 따라 회귀 또는 분류 모델을 적절히 선택해야 함
    - 가령 시험점수를 예측하는 것과 같은 부분은 회귀모델을 시점 점수에 따른 합격/불합격 여부를 예측하는 것은 분류 모델을 적용
    - 회귀는 양적 데이터, 즉 점수와 같이 연속된 어떠한 값을 예측하는 모델, 분류는 합격/불합격과 같은 범주형 데이터 즉, 어떤 종류를 예측하는 모델, 분류는 이진 분류, 다중 분류로 구분 가능

- 비지도학습 이해하기
    - 비지도학습(Unsupervised Learning)은 정답이 없는 상태에서 입력 데이터만 가지고 학습하는 방법, 정답 즉 레이블이 없기 때문에 구체적인 결과가 아니라 군집화(Clustering)를 예측할 때 주로 사용
    - 군집화란 주어진 데ㅣ터를 기반으로 그 값이 유사한 데이터들의 그룹을 나누는 것, 군집화 알고리즘 중 대표적인 k-means 군집화 에서 k란 입력된 데이터를 몇 개의 그룹으로 나누느녀 하는 것, k 값은 사용자가 선택해야 하며 k 개 군집의 중심점부터 가장 가까운 데이터들을 그룹핑한 결과
