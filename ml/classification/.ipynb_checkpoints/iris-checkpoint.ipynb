```python
import pandas as pd
```

```python
iris_pd = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_pd.head()
```

```python
iris_pd["species"] = iris.target
iris_pd.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x="sepal length (cm)", y="species", hue="species", data=iris_pd, orient="h")
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x="sepal width (cm)", y="species", hue="species", data=iris_pd, orient="h")
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x="petal length (cm)", y="species", hue="species", data=iris_pd, orient="h")
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x="petal width (cm)", y="species", hue="species", data=iris_pd, orient="h")
```

```python
sns.pairplot(iris_pd, hue="species")
```

```python
sns.pairplot(
    data=iris_pd,
    vars=["petal length (cm)", "petal width (cm)"],
    hue="species",
    height=4,
)
```

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(
    x="petal length (cm)",
    y="petal width (cm)",
    data=iris_pd,
    hue="species",
    palette="Set2",
)
```

```python
iris_12 = iris_pd[iris_pd["species"] != 0]
iris_12.info()
```

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(
    x="petal length (cm)",
    y="petal width (cm)",
    data=iris_12,
    hue="species",
    palette="Set2",
)
```

```python
from sklearn.tree import DecisionTreeClassifier

iris_tree = DecisionTreeClassifier()
iris_tree.fit(iris.data[:, 2:], iris.target)
```

```python
from sklearn.metrics import accuracy_score

y_pred_tr = iris_tree.predict(iris.data[:, 2:])
accuracy_score(iris.target, y_pred_tr)
```
