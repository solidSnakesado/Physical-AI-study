- 학습 순서
    - 평균값 필터 -> 가우시안 필터 -> 샤프닝 -> 잡음 제거 -> 응용 필터 순으로 진행
    - 각 필터마다 원리(그림) -> 함수 시그니처 -> 실습 코드 -> 실행 결과 4단계 반복
    - 실습 영상: rose.bmp(부드럽게/샤프닝), noise.bmp(잡음 제거), lenna.bmp(양방향 필터)

- 영상 필터링이란
    - 핵심 개념
        - 필터링(filtering: 영상에서 원하는 정보만 통과시키고 나머지는 걸러 내는 연산
        - 마스크(mask)/커널(kernel): 필터의 모양/계수를 담은 작은 행렬. 보톤 3x3, 5x5 등 홀수 크기 정사각형
        - 합성곱(convolution): 마스크를 영상 위에서 한 픽셀씩 이동시키며 겹친 값까지 곱해 더한 결과를 중앙 픽셀에 기록
        - 고정점(anchor): 결과를 저장할 기준 위치. 보통 마스크 중앙
        - 가장자리 처리(borderType): 마스크가 영상 밖으로 나가는 구간을 어떻게 채울지 결정

- 필터의 두 갈래
    - 블러(저역 통과): 인접 픽셀 값을 섞어 부드럽게 함 -> 평균값/가우시안/메디안/양방향 필터
    - 샤프닝(고역 강조): 부드러운 성분을 빼서 경계를 도드라지게 함 -> 언샤프 마스크
