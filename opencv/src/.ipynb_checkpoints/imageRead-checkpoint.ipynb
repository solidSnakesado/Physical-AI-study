```python
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("../data/lenna.bmp")
type(img)
```

```python
img.shape
```

```python
plt.imshow(img)
```

- 그림이 파란색 필터가 있는 것처럼 표시
- matplotlib는 rgb 순서로 이미지를 읽는 데 opencv는 bgr 순서로 읽기 때문
- cvtColor에 COLOR_BGR2RGB를 적용

```python
fix_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(fix_img)
```

```python
img_gray = cv2.imread("../data/lenna.bmp", cv2.IMREAD_GRAYSCALE)
plt.imshow(img_gray, cmap="gray")
```
